## 1. Importar Pandas

`import pandas as pd` importa la librería y le asigna el alias  `pd`.


```python
import pandas as pd
```

**Resultado:** Pandas queda disponible como `pd` para trabajar con nuestra base.

## 2. Cargar una base de datos

`pd.read_csv("archivo.csv")` lee un CSV. Guardamos el resultado en la variable `df` para reutilizarlo.

```python
df = pd.read_csv("base_generica_pandas.csv", encoding="utf-8")
df.head()
```

**Resultado:** `df` contiene los 100 registros y las 10 columnas del CSV; se muestran las primeras cinco filas. El índice de la izquierda identifica las filas y no es una columna del archivo.

## 3. Conocer nuestro DataFrame

Estos métodos y atributos permiten revisar la estructura antes de analizar los datos.

### Primeras y últimas filas

`head()` muestra las primeras cinco filas y `tail()` las últimas cinco. Podemos indicar otra cantidad con `head(n)` o `tail(n)`.

```python
display(df.head())
display(df.tail(8))
```

**Resultado:** la primera tabla muestra las filas con índices del 0 al 4; la segunda, las ocho últimas, con índices del 92 al 99.

### Dimensiones y nombres

`shape` y `columns` son atributos: se escriben **sin paréntesis**. `shape` devuelve `(filas, columnas)`.

```python
print("Dimensiones:", df.shape)
print("Columnas:", df.columns.tolist())
```

**Resultado:** `(100, 10)` indica 100 filas y 10 columnas. La lista contiene los nombres desde `col1` hasta `col10`.

### Información general

`info()` imprime los nombres de las columnas, la cantidad de valores **no nulos**, y los tipos de datos 

```python
df.info()
```

**Resultado:** cada columna tiene 100 valores no nulos; no hay datos faltantes. Se distinguen columnas numéricas y de texto.

## 4. Seleccionar columnas

Escribimos el nombre de cada columna entre comillas. Un nombre selecciona una `Series`; una lista de nombres selecciona un `DataFrame`.

### Una columna

Sintaxis: `df["columna"]`. Usamos `head()` para mantener breve la vista del ejemplo.

```python
df["col5"].head()
```

**Resultado:** una `Series` con las primeras cinco categorías de `col5`. Sin `head()`, la selección incluye los 100 registros.

### Varias columnas

Sintaxis: `df[["columna1", "columna2"]]`. El corchete interior crea una lista.

```python
df[["col8", "col7"]].head()
```

**Resultado:** un `DataFrame` con las primeras cinco filas de `col8` y `col7`, en ese orden.

### Selección mediante listas

Podemos guardar los nombres en una lista y pasarla a `df[...]`.

```python
columnas_seleccionadas = ["col1", "col10", "col6", "col8"]
df[columnas_seleccionadas].head()
```

**Resultado:** una tabla con las cuatro columnas solicitadas y una vista de cinco filas.

## 5. Filtrar datos

Una comparación como `df["col6"] == "Sur"` produce valores `True` y `False`. `df[condicion]` conserva solo las filas donde la condición es verdadera.

### Condición numérica

Usamos operadores como `>`,  `>=`,  y  `!=`.

```python
df[df["col1"] > 500]
```

**Resultado:** solo los registros cuyo valor de `col1` es mayor que 500; el valor 500 queda excluido.

### Condición categórica

Para comparar texto, escribimos el valor entre comillas. `==` compara; `=` asigna.

```python
df[df["col6"] == "Sur"]
```

**Resultado:** los 24 registros de la categoría `Sur` en `col6`.

### Varias condiciones: 

Encerramos cada comparación entre paréntesis y usamos `&` para exigir que ambas se cumplan.

```python
df[(df["col6"] == "Sur") & (df["col5"] != "B")]
```

**Resultado:** filas de `Sur` cuya categoría en `col5` es diferente de `B`.

### Varias condiciones: O y negación

Usamos `|` para que se cumpla al menos una condición y `~` para negar una máscara. Para combinar Series usamos `&` y `|`

```python
display(df[(df["col5"] == "A") | (df["col6"] == "Norte")])
display(df[~(df["col6"] == "Sur")].head())
```

**Resultado:** la primera tabla incluye filas con `col5` igual a `A`, con región `Norte` o con ambas condiciones. La segunda muestra cinco registros que no pertenecen a `Sur`.

## 6. Ordenar datos

`sort_values()` ordena por los valores de una o varias columnas. Por defecto, el orden es ascendente. 

### Una columna

`ascending=False` cambia el orden a descendente.

```python
display(df.sort_values("col1").head())
display(df.sort_values("col1", ascending=False).head())
```

**Resultado:** se muestran los cinco valores más pequeños de `col1` y luego los cinco más grandes, junto con sus demás columnas.

### Varias columnas

Pasamos listas a `by` y a `ascending`: cada dirección corresponde a una columna.

```python
df.sort_values(by=["col6", "col1"], ascending=[True, False]).head(10)
```

**Resultado:** filas ordenadas alfabéticamente por `col6` y, dentro de cada región, por `col1` de mayor a menor. El índice original se conserva.

## 7. Estadística descriptiva

Estas operaciones resumen los valores numéricos. Normalmente omiten los valores faltantes

### Resumen general

`describe()` calcula conteo, media, desviación estándar, mínimo, cuartiles y máximo de las columnas numéricas.

```python
df.describe()
```

**Resultado:** una tabla con ocho estadísticas por columna numérica. `50%` es la mediana; `25%` y `75%` son los otros cuartiles.

### Estadísticas de una columna

`mean()` calcula la media; `median()`, la mediana; `min()` y `max()`, los extremos; `std()`, la desviación estándar muestral (por defecto, `ddof=1`).

```python
print("Media:", df["col1"].mean())
print("Mediana:", df["col1"].median())
print("Mínimo:", df["col1"].min())
print("Máximo:", df["col1"].max())
print("Desviación estándar:", df["col1"].std())
```

**Resultado:** para `col1`, la media es aproximadamente 475.48, la mediana 469.50, el mínimo 15.47, el máximo 987.02 y la desviación estándar 294.51.

## 8. Trabajar con variables categóricas

Podemos consultar qué categorías existen y cuántas veces aparecen, sin realizar operaciones aritméticas con ellas.

### Frecuencias

`value_counts()` cuenta los registros de cada categoría y ordena por frecuencia descendente.

```python
df["col6"].value_counts()
```

**Resultado:** `Norte` aparece 33 veces; `Sur`, 24; `Oeste`, 22; y `Este`, 21. Por defecto, los nulos se excluyen; `dropna=False` permite incluirlos.

### Categorías distintas

`unique()` devuelve los valores distintos en orden de aparición; `nunique()` devuelve cuántos hay, excluyendo nulos por defecto.

```python
print("Categorías:", df["col6"].unique())
print("Número de categorías:", df["col6"].nunique())
```

**Resultado:** los valores distintos son `Sur`, `Oeste`, `Norte` y `Este`

## 9. Agrupar datos

`groupby()` separa los registros por categorías para calcular estadísticas dentro de cada grupo. Después de agrupar elegimos una operación como `mean()`, `size()` o `agg()`.

### Promedio por categoría

Seleccionamos `col1` después de agrupar por `col10` y calculamos su media.

```python
df.groupby("col10")["col1"].mean()
```

**Resultado:** una media de `col1` para cada categoría de `col10`: `Alto`, `Bajo` y `Medio`.

### Conteos por categoría

`size()` cuenta filas, mientras que `count()` cuenta valores no nulos de cada columna seleccionada. También podemos agrupar por más de una columna.

```python
display(df.groupby("col6").size())
display(df.groupby(["col10", "col5"])[["col1", "col3"]].count())
```

**Resultado:** la primera salida cuenta registros por región. La segunda cuenta valores no nulos de `col1` y `col3` por combinación de `col10` y `col5`; ambos conteos coinciden porque no hay nulos.

### Varias agregaciones

`agg()` permite calcular varias estadísticas en una sola operación. Cada par indica la columna de origen y la función.

```python
resumen = df.groupby("col6").agg(
    registros=("col1", "size"),
    promedio_col1=("col1", "mean"),
    minimo_col1=("col1", "min"),
    maximo_col1=("col1", "max")
)
resumen
```

**Resultado:** una tabla de cuatro filas, una por región, con el número de registros y tres estadísticas de `col1`.

## 10. Crear y modificar columnas

`df["nueva_columna"] = expresion` crea una columna; si el nombre ya existe, reemplaza sus valores. Las operaciones entre columnas se realizan fila por fila. Usaremos `df_editado = df.copy()` para practicar y poder repetir la sección desde la base original.

### Crear columnas

Creamos una columna numérica y otra lógica a partir de `col1`.

```python
df_editado = df.copy()
df_editado["col1_doble"] = df_editado["col1"] * 2
df_editado["supera_500"] = df_editado["col1"] > 500
df_editado[["col1", "col1_doble", "supera_500"]].head()
```

**Resultado:** dos nuevas columnas: el doble de `col1` y un indicador `True` o `False` que señala si supera 500.

### Modificar una columna y asignar con condición

Podemos reemplazar una columna completa o usar `.loc[condicion, "columna"]` para modificar solo algunas filas.

```python
df_editado["col1_doble"] = df_editado["col1_doble"].round(1)
df_editado.loc[df_editado["col6"] == "Sur", "col9"] = "Sí"
df_editado.loc[df_editado["col6"] == "Sur", ["col6", "col9", "col1_doble"]].head()
```

**Resultado:** `col1_doble` queda redondeada a un decimal y `col9` vale `Sí` en todas las filas de `Sur` de la copia. La vista muestra cinco de esas filas. El `df` original conserva sus valores.
